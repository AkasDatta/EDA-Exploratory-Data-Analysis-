```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('titanic_data_updated.csv')
```

```python
df.shape

print(f"Total number of rows: {df.shape[0]}")
print(f"Total number of columns: {df.shape[1]}")
```

```python
df.head(10)  #first 10 rows of the dataset 
df.sample(10) #random 10 rows of the dataset 
```

```python
# missing 
df.isnull().sum()
```

```python
# checking duplicates 
df.duplicated().sum()

# duplicate value remove
df.drop_duplicates(inplace=True)
```

```python
df.info()
```

```python
df.describe()
```

```python
sns.countplot(data = df, x = df["Survived"])
```

```python
df["Survived"].value_counts()
```

```python
df["Survived"].value_counts() / len(df) * 100
```

```python
survived_count = df["Survived"].value_counts()
plt.pie(survived_count)
```

```python
survived_count = df["Survived"].value_counts()
plt.pie(survived_count, autopct="%1.1f")
```

```python
survived_count = df["Survived"].value_counts()

survived_label = df["Survived"].unique()
print(survived_label)

plt.pie(survived_count, labels=survived_label, autopct="%1.1f")
```

```python
survived_count = df["Survived"].value_counts()

survived_label = df["Survived"].unique()
print(survived_label)

plt.pie(survived_count, labels=survived_label, autopct="%1.1f")

plt.title("Pie chart showing")
plt.show()
```
